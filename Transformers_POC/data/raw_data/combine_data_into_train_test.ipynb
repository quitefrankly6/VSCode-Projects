```python
import pandas as pd 
import numpy as np
import os
from pathlib import Path
```

```python
directory_path = Path('.') # Create a Path object

# List all files in the directory (non-recursive)
file_list_pathlib = [str(file) for file in directory_path.iterdir() if file.is_file() and '.csv' in str(file)]

print(file_list_pathlib)
```

```python
df_dict  = {}

for file in file_list_pathlib:
    try:
        # Read the file and use the filename as the dictionary key
        df = pd.read_csv(file)
        # os.path.basename(file) gets just the filename without the full path
        df_dict[os.path.basename(file)] = df
        print(f"Successfully read {file}")
    except Exception as e:
        print(f"Error reading {file}: {e}")
        
```

```python
cols_to_select = ['Description', 'Source', 'Rent']
```

```python
selected_dfs = []
```

```python
_to_append = df_dict['Checking1.csv']
_to_append['Source'] = 'real' 

selected_dfs.append(_to_append[cols_to_select])
```

```python
_to_append = df_dict['frank_labeled_data.csv'].rename(columns = {'rent':'Rent'})
_to_append['Source'] = 'real' 

selected_dfs.append(_to_append[cols_to_select])
```

```python
_to_append = df_dict['deng_transactions.csv']
_to_append['Rent'] = 0
_to_append['Source'] = 'real' 

selected_dfs.append(_to_append[cols_to_select])
```

```python
_to_append = df_dict['bank_statement.csv']
_to_append['Source'] = 'real' 

selected_dfs.append(_to_append[cols_to_select])
```

```python
_to_append = df_dict['rent_dataset_hard_1000.csv'].rename(columns = {'rent':'Rent'})
_to_append['Source'] = 'synthetic' 

selected_dfs.append(_to_append[cols_to_select])
```

```python
selected_dfs
```

```python
# Combine all DataFrames into a single one
if selected_dfs:
    combined_df = pd.concat(selected_dfs, ignore_index=True)
    print("\nSuccessfully combined all files into a single DataFrame.")
    print(f"Total combined shape: {combined_df.shape}")
    # print the first few rows of the combined dataframe
    print(combined_df.head()) 
else:
    print("No dataframes were successfully loaded to combine.")
```

```python
combined_df.value_counts(['Source', 'Rent'])
```

```python
combined_df.to_csv('../processed_data/all_data.csv')
```
